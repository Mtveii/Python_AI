```python
low = int(input(""))
high = int(input(""))
```

```python
matrix = np.random.randint(low, high + 1, size=(5, 5))
print(matrix)
```

```python
diag = np.diag(matrix)
print(diag)
```

```python
diag_sum = np.sum(diag)
print(diag_sum)
```
